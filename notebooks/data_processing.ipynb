## Traitement et Sauvegarde

Le code complet de `DataProcessor` est dans `src/data_processing.py`. Nous l'importons directement pour éviter la redondance.


```python
# Traitement complet des données
processor = DataProcessor(data_dir="../data/raw")

# Charger les données
train_df, test_df = processor.load_data()

print(f"Données d'entraînement: {train_df.shape}")
print(f"Données de test: {test_df.shape}")

# Traiter les valeurs manquantes
train_df_cleaned = processor.handle_missing_values(train_df, is_train=True)
test_df_cleaned = processor.handle_missing_values(test_df, is_train=False)

print("\nVérification des valeurs manquantes après traitement:")
print(f"Train - Valeurs manquantes: {train_df_cleaned.isnull().sum().sum()}")
print(f"Test - Valeurs manquantes: {test_df_cleaned.isnull().sum().sum()}")

# Transformer les outliers (méthode contextuelle)
train_df_final = processor.transform_outliers(train_df_cleaned, target_col='SalePrice', method='contextual')

print(f"\nDonnées finales d'entraînement: {train_df_final.shape}")

# Sauvegarder les données traitées
output_dir = Path("../data/processed")
output_dir.mkdir(parents=True, exist_ok=True)

train_output_path = output_dir / "train_cleaned.csv"
test_output_path = output_dir / "test_cleaned.csv"

train_df_final.to_csv(train_output_path, index=False)
test_df_cleaned.to_csv(test_output_path, index=False)

print(f"\n✅ Données sauvegardées avec succès:")
print(f"   - {train_output_path}")
print(f"   - {test_output_path}")
print(f"\nCes fichiers peuvent maintenant être utilisés dans les autres notebooks!")
```
