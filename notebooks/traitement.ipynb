## 1) Audit des valeurs manquantes (NA)

Objectif :
- Identifier les variables contenant des valeurs manquantes dans le jeu **train**
- Comprendre la nature des NA :
  - **NA structurels** : absence réelle d’un équipement (piscine, garage, sous-sol…)
  - **NA non structurels** : information attendue mais manquante (ex : LotFrontage)

Dans cette étape, on ne fait PAS encore :
- d’imputation statistique (médiane, KNN, modèles…)
- d’encodage (one-hot, ordinal encoding…)
- de normalisation / scaling

On fait uniquement un nettoyage "métier" des NA structurels.

OCCUPONS NOUS D'ABORD DE LA BASE TRAIN VU QUE ELLE ET LA BASE TEST PEUVENT SE COMPORTER DIFFEREMMENT AVEC LES VARIABLES A NA

```python
# Audit initial des valeurs manquantes dans le jeu d'entraînement
# Objectif : identifier toutes les variables contenant au moins une valeur manquante

# Calcul du nombre de valeurs manquantes par variable
na_counts = train.isna().sum()

# Filtrage : ne conserver que les variables avec au moins une valeur manquante
# Tri décroissant pour identifier les variables les plus problématiques
na_counts_nonzero = na_counts[na_counts > 0].sort_values(ascending=False)

print(f"Nombre de variables avec au moins 1 NA : {na_counts_nonzero.shape[0]}")
na_counts_nonzero
```

```python
# Calcul du pourcentage de valeurs manquantes par variable
# Cette métrique permet d'évaluer l'ampleur du problème pour chaque variable
na_percent = (na_counts_nonzero / len(train) * 100).round(2)

# Création d'un tableau récapitulatif combinant nombre et pourcentage de NA
# Tri par nombre de valeurs manquantes décroissant pour faciliter l'analyse
na_summary = (
    pd.DataFrame({"na_count": na_counts_nonzero, "na_percent": na_percent})
    .sort_values("na_count", ascending=False)
)

na_summary
```

Une fois qu'on a la liste des variables ayant des NA, on va détecter dans cette liste les NA structurels (les NA qui le sont du fait de l'absence de matériel)

```python
# Identification des variables à valeurs manquantes structurelles
# Les NA structurels représentent l'absence réelle d'un équipement ou d'une caractéristique
# (exemple : pas de piscine, pas de garage, pas de sous-sol)
# Ces valeurs manquantes ne sont pas des erreurs mais des informations valides

na_structural_vars = [
    "PoolQC", "MiscFeature", "Alley", "Fence",
    "FireplaceQu",
    "GarageType", "GarageYrBlt", "GarageFinish", "GarageQual", "GarageCond",
    "BsmtExposure", "BsmtFinType2", "BsmtQual", "BsmtCond", "BsmtFinType1", "MasVnrType"
]
# Note : MasVnrType est une variable à NA structurels mais son traitement est reporté
# car elle est liée à MasVnrArea qui nécessite un traitement spécifique

# Vérification de cohérence : contrôle que toutes les variables sont correctement classées
missing_in_list = set(na_summary.index) - set(na_structural_vars)  # Variables avec NA non classées comme structurelles
extra_in_list = set(na_structural_vars) - set(na_summary.index)    # Variables classées structurelles mais sans NA dans train

print("NA présents mais NON classés structurels :", sorted(missing_in_list))
print("Classés structurels mais sans NA dans train :", sorted(extra_in_list))
```

```python
# Identification des variables à valeurs manquantes non structurelles
# Les NA non structurels représentent des informations attendues mais manquantes
# Ces valeurs nécessitent une imputation statistique (médiane, mode, etc.)
na_non_structural_vars = sorted(list(missing_in_list))

na_non_structural_vars

```

```python
# Séparation des variables structurelles en qualitatives et numériques
# Les variables qualitatives structurelles seront remplacées par "None" (chaîne de caractères)
# Les variables numériques structurelles nécessitent un traitement spécifique

na_structural_cat = [v for v in na_structural_vars if v != "GarageYrBlt"]

# Cas particulier : GarageYrBlt est une variable numérique structurelle
# Son traitement sera effectué séparément avec création d'un indicateur binaire
print("NA structurels qualitatifs :", na_structural_cat)
print("NA structurel numérique (à part) :", ["GarageYrBlt"])
```

```python
# Création d'une copie du jeu d'entraînement pour le nettoyage
# Important : ne jamais modifier directement les données brutes pour préserver l'intégrité des données originales
train_clean = train.copy()

# =========================
# Traitement des valeurs manquantes structurelles qualitatives
# =========================
# Remplacement des NA structurels par la chaîne "None" pour indiquer explicitement l'absence
# Cette approche transforme les valeurs manquantes en modalité catégorielle valide
for col in na_structural_cat:
    train_clean[col] = train_clean[col].fillna("None")

# Vérification : contrôle qu'il ne reste plus de valeurs manquantes sur ces colonnes
train_clean[na_structural_cat].isna().sum().sort_values(ascending=False).head(10)

```

```python
# =========================
# Vérification : confirmation que "None" est bien devenu une modalité valide
# =========================
# Cette vérification permet de s'assurer que le remplacement des NA par "None"
# a bien fonctionné et que "None" apparaît comme une catégorie dans les distributions
cols_check = ["PoolQC", "Fence", "GarageType", "BsmtQual"]
for c in cols_check:
    if c in train_clean.columns:
        print(f"\n{c}")
        print(train_clean[c].value_counts(dropna=False).head(10))
```

```python
# =========================
# État des valeurs manquantes après traitement des NA structurels
# =========================
# Identification des variables qui contiennent encore des valeurs manquantes
# après le traitement des NA structurels qualitatifs
na_counts = train_clean.isna().sum()
na_counts[na_counts > 0].sort_values(ascending=False)
```

```python
# =========================
# Traitement de MasVnrArea (surface du parement en maçonnerie)
# =========================
# MasVnrType : type de parement (BrkFace, Stone, etc.) - variable qualitative
# MasVnrArea : surface du parement - variable numérique
# Relation logique : si MasVnrType est "None" (pas de parement), alors MasVnrArea doit être 0
# Imputation : remplacer les NA de MasVnrArea par 0 (cohérent avec l'absence de parement)

if "MasVnrArea" in train_clean.columns:
    train_clean["MasVnrArea"] = train_clean["MasVnrArea"].fillna(0)

# Vérification de cohérence entre MasVnrType et MasVnrArea
train_clean[["MasVnrType", "MasVnrArea"]].isna().sum()
```

```python
# =========================
# Imputation de la variable Electrical
# =========================
# Variable non structurelle avec une seule valeur manquante
# Stratégie : imputation par la modalité la plus fréquente (mode)
# Cette approche est appropriée pour une variable catégorielle avec très peu de valeurs manquantes

if "Electrical" in train_clean.columns:
    mode_elec = train_clean["Electrical"].mode()[0]
    train_clean["Electrical"] = train_clean["Electrical"].fillna(mode_elec)

    print("Mode utilisé pour Electrical :", mode_elec)
    print("NA restants Electrical :", train_clean["Electrical"].isna().sum())
```

```python
# =========================
# Traitement de GarageYrBlt (année de construction du garage)
# =========================
# Variable numérique structurelle nécessitant un traitement en deux étapes :
# 1. Création d'un indicateur binaire pour distinguer présence/absence de garage
# 2. Imputation conditionnelle selon la présence ou l'absence de garage

# Création de l'indicateur binaire HasGarage
# Valeur 1 si garage présent, 0 sinon
train_clean["HasGarage"] = (train_clean["GarageType"] != "None").astype(int)

# Imputation conditionnelle de GarageYrBlt
if "GarageYrBlt" in train_clean.columns:
    # Cas 1 : pas de garage -> valeur sentinelle 0
    train_clean.loc[train_clean["HasGarage"] == 0, "GarageYrBlt"] = train_clean.loc[train_clean["HasGarage"] == 0, "GarageYrBlt"].fillna(0)

    # Cas 2 : garage présent mais année manquante -> médiane des maisons avec garage
    # Cette imputation est plus réaliste que la médiane globale
    median_garage_year = train_clean.loc[train_clean["HasGarage"] == 1, "GarageYrBlt"].median()
    train_clean.loc[train_clean["HasGarage"] == 1, "GarageYrBlt"] = train_clean.loc[train_clean["HasGarage"] == 1, "GarageYrBlt"].fillna(median_garage_year)

    print("Médiane GarageYrBlt (maisons avec garage) :", median_garage_year)
    print("NA restants GarageYrBlt :", train_clean["GarageYrBlt"].isna().sum())
```

```python
# =========================
# Imputation de LotFrontage (largeur de façade du lot)
# =========================
# Variable non structurelle : information attendue mais manquante
# Stratégie : imputation hiérarchique par quartier (Neighborhood)
# Justification : la largeur de façade est corrélée au quartier (zones résidentielles similaires)

if "LotFrontage" in train_clean.columns and "Neighborhood" in train_clean.columns:
    # Étape 1 : imputation par la médiane du quartier
    # Cette approche préserve la cohérence spatiale des données
    train_clean["LotFrontage"] = train_clean["LotFrontage"].fillna(
        train_clean.groupby("Neighborhood")["LotFrontage"].transform("median")
    )
    # Étape 2 : fallback sur la médiane globale pour les cas où le quartier n'a pas de données
    # Cette étape garantit qu'aucune valeur manquante ne subsiste
    train_clean["LotFrontage"] = train_clean["LotFrontage"].fillna(train_clean["LotFrontage"].median())

    print("NA restants LotFrontage :", train_clean["LotFrontage"].isna().sum())
```

```python
# =========================
# Contrôle final : vérification de l'absence de valeurs manquantes
# =========================
# Cette vérification confirme que tous les traitements ont été effectués correctement
# et qu'aucune valeur manquante ne subsiste dans le jeu d'entraînement nettoyé
na_final = train_clean.isna().sum()
na_final[na_final > 0].sort_values(ascending=False)
```

```python
## Traitement des valeurs manquantes dans le jeu de test
# =====================================================
# Le jeu de test peut présenter des valeurs manquantes différentes du jeu d'entraînement
# Il est essentiel d'appliquer les mêmes stratégies de traitement pour garantir la cohérence

# Audit initial des valeurs manquantes dans le jeu de test
na_counts_test = test.isna().sum()

# Identification des variables avec au moins une valeur manquante
na_counts_nonzero_test = na_counts_test[na_counts_test > 0].sort_values(ascending=False)

print(f"Nombre de variables avec au moins 1 NA : {na_counts_nonzero_test.shape[0]}")
na_counts_nonzero_test
```

```python
# Calcul du pourcentage de valeurs manquantes dans le jeu de test
# Note : le dénominateur utilise len(train) pour permettre la comparaison avec le jeu d'entraînement
na_percent_test = (na_counts_nonzero_test / len(train) * 100).round(2)

# Création d'un tableau récapitulatif pour le jeu de test
na_summary_test = (
    pd.DataFrame({"na_count": na_counts_nonzero_test, "na_percent": na_percent_test})
    .sort_values("na_count", ascending=False)
)

na_summary_test
```

```python
# ============================
# Classification des variables pour le traitement du jeu de test
# ============================
# Cette classification permet d'appliquer les mêmes stratégies de traitement
# que celles utilisées pour le jeu d'entraînement, garantissant la cohérence

# Variables structurelles d'existence (qualitatives principales)
# Ces variables indiquent la présence ou l'absence d'un équipement
structural_exist_vars = [
    "PoolQC",        # Piscine
    "MiscFeature",  # Équipements divers
    "Alley",         # Accès / extérieur
    "Fence",         # Clôture
    "FireplaceQu",   # Cheminée
    "GarageType",    # Garage
    "BsmtQual",      # Sous-sol
    "MasVnrType"     # Parement
]

# Variables structurelles qualitatives dépendantes
# Ces variables décrivent des caractéristiques d'équipements existants
structural_qual_dependent_vars = [
    "GarageFinish",  # Garage
    "GarageQual",    # Garage
    "GarageCond",    # Garage
    "BsmtCond",      # Sous-sol
    "BsmtExposure",  # Sous-sol
    "BsmtFinType1",  # Sous-sol
    "BsmtFinType2"   # Sous-sol
]

# Variables structurelles numériques dépendantes
# Ces variables quantifient des caractéristiques d'équipements existants
structural_num_dependent_vars = [
    "GarageYrBlt",   # Garage
    "GarageArea",    # Garage
    "GarageCars",    # Garage
    "TotalBsmtSF",   # Sous-sol
    "BsmtUnfSF",     # Sous-sol
    "BsmtFinSF1",    # Sous-sol
    "BsmtFinSF2",    # Sous-sol
    "BsmtFullBath",  # Sous-sol
    "BsmtHalfBath",  # Sous-sol
    "MasVnrArea"     # Parement
]

# Variables non structurelles
# Ces variables nécessitent une imputation statistique
non_structural_vars = [
    "LotFrontage",
    "Electrical",
    "MSZoning",
    "Functional",
    "Utilities",
    "Exterior1st",
    "Exterior2nd",
    "KitchenQual",
    "SaleType"
]
```

```python
# Création d'une copie du jeu de test pour le nettoyage
# Important : préserver l'intégrité des données brutes en travaillant sur une copie
test_clean = test.copy()
```

```python
# ============================
# B) Variables structurelles d’existence
# ============================

for col in structural_exist_vars:
    if col in test_clean.columns:
        test_clean[col] = test_clean[col].fillna("None")

```

```python
# ============================
# Traitement des variables structurelles qualitatives dépendantes
# ============================
# Remplacement des valeurs manquantes par "None" pour les caractéristiques d'équipements absents
# Ces variables dépendent des variables d'existence traitées précédemment

for col in structural_qual_dependent_vars:
    if col in test_clean.columns:
        test_clean[col] = test_clean[col].fillna("None")
```

```python
# ============================
# D1) Indicateurs d’existence
# ============================

# Garage
test_clean["HasGarage"] = (test_clean["GarageType"] != "None").astype(int)

# Sous-sol
test_clean["HasBsmt"] = (test_clean["BsmtQual"] != "None").astype(int)

# Parement
test_clean["HasMasVnr"] = (test_clean["MasVnrType"] != "None").astype(int)
```

```python
# ============================
# Imputations conditionnelles des variables numériques dépendantes
# ============================
# Principe : si l'équipement est absent, les variables numériques associées sont mises à 0
# Cette approche garantit la cohérence logique entre les variables qualitatives et numériques

# Variables numériques liées au garage
garage_num_vars = ["GarageYrBlt", "GarageArea", "GarageCars"]
for col in garage_num_vars:
    if col in test_clean.columns:
        # Si pas de garage, mettre les valeurs numériques à 0
        test_clean.loc[test_clean["HasGarage"] == 0, col] = 0

# Variables numériques liées au sous-sol
bsmt_num_vars = [
    "TotalBsmtSF", "BsmtUnfSF", "BsmtFinSF1", "BsmtFinSF2",
    "BsmtFullBath", "BsmtHalfBath"
]
for col in bsmt_num_vars:
    if col in test_clean.columns:
        # Si pas de sous-sol, mettre les valeurs numériques à 0
        test_clean.loc[test_clean["HasBsmt"] == 0, col] = 0

# Variable numérique liée au parement
if "MasVnrArea" in test_clean.columns:
    # Si pas de parement, mettre la surface à 0
    test_clean.loc[test_clean["HasMasVnr"] == 0, "MasVnrArea"] = 0
```

```python
# ============================
# Imputation de GarageYrBlt pour les garages existants avec année manquante
# ============================
# Cas particulier : garage présent mais année de construction manquante
# Stratégie : utiliser la médiane calculée sur le jeu d'entraînement
# Cette approche garantit la cohérence entre train et test

if "GarageYrBlt" in test_clean.columns:
    test_clean.loc[
        (test_clean["HasGarage"] == 1) & (test_clean["GarageYrBlt"].isna()),
        "GarageYrBlt"
    ] = median_garage_year  # Médiane calculée sur le jeu d'entraînement
```

```python
# ============================
# Vérification 1 : État des valeurs manquantes après traitement structurel
# ============================
# Cette vérification permet d'identifier les variables nécessitant encore un traitement
# et de quantifier l'ampleur du travail restant

na_total = test_clean.isna().sum().sum()
print("Nombre TOTAL de valeurs manquantes (NA) dans test_clean :", na_total)

na_by_col = test_clean.isna().sum()
na_cols = na_by_col[na_by_col > 0].sort_values(ascending=False)

print("\nNombre de colonnes avec au moins 1 NA :", na_cols.shape[0])
na_cols
```

```python
# ============================
# Vérification 2 : Comparaison avant/après traitement structurel
# ============================
# Cette analyse permet d'évaluer l'efficacité du traitement et d'identifier
# les variables qui nécessitent encore une imputation non structurelle

na_after = test_clean.isna().sum()
before_cols = set(na_counts_nonzero_test.index)  # Variables avec NA avant traitement
after_cols  = set(na_after[na_after > 0].index) # Variables avec NA après traitement

print("Colonnes qui avaient des NA avant :", len(before_cols))
print("Colonnes qui ont encore des NA après :", len(after_cols))

print("\n NA réglés (présents avant, absents après) :", sorted(list(before_cols - after_cols)))
print("NA restants (encore après nettoyage) :", sorted(list(after_cols)))
```

```python
# ============================
# Vérification 3 : Confirmation de la présence de "None" comme modalité
# ============================
# Cette vérification permet de s'assurer que le remplacement des NA structurels
# par "None" a bien fonctionné et que "None" apparaît dans les distributions

cols_check = ["PoolQC", "Fence", "GarageType", "BsmtQual", "MasVnrType"]
for c in cols_check:
    if c in test_clean.columns:
        print(f"\n--- {c} ---")
        print("NA :", test_clean[c].isna().sum())
        print(test_clean[c].value_counts(dropna=False).head(10))
```

```python
# ============================
# Vérification 4 : Cohérence logique entre MasVnrType et MasVnrArea
# ============================
# Cette vérification permet de détecter d'éventuelles incohérences logiques :
# - MasVnrType = "None" mais MasVnrArea > 0 (incohérence)
# - MasVnrType != "None" mais MasVnrArea = 0 (possible mais à vérifier)

if "MasVnrType" in test_clean.columns and "MasVnrArea" in test_clean.columns:
    print("NA MasVnrType :", test_clean["MasVnrType"].isna().sum())
    print("NA MasVnrArea :", test_clean["MasVnrArea"].isna().sum())

    # Détection des incohérences potentielles
    incoh_1 = test_clean[(test_clean["MasVnrType"] == "None") & (test_clean["MasVnrArea"] > 0)].shape[0]
    incoh_2 = test_clean[(test_clean["MasVnrType"] != "None") & (test_clean["MasVnrArea"] == 0)].shape[0]


```

```python
# ============================
# Imputation des valeurs manquantes non structurelles dans le jeu de test
# ============================
# Stratégie : utiliser les statistiques calculées sur le jeu d'entraînement
# pour garantir la cohérence et éviter le data leakage

# A) Variables qualitatives : imputation par le mode du jeu d'entraînement
cat_non_structural = [
    "MSZoning",
    "Utilities",
    "Functional",
    "Exterior1st",
    "Exterior2nd",
    "KitchenQual",
    "SaleType"
]

for col in cat_non_structural:
    if col in test_clean.columns:
        # Utiliser le mode du jeu d'entraînement pour l'imputation
        mode_col = train_clean[col].mode()[0]
        test_clean[col] = test_clean[col].fillna(mode_col)

# B) Variables numériques : imputation conditionnelle pour les garages existants
# Cas : garage présent mais information manquante (rare mais possible)
num_non_structural = ["GarageCars", "GarageArea"]

for col in num_non_structural:
    if col in test_clean.columns:
        # Utiliser la médiane des maisons avec garage du jeu d'entraînement
        median_col = train_clean.loc[train_clean["HasGarage"] == 1, col].median()
        test_clean[col] = test_clean[col].fillna(median_col)
```

```python
# ============================
# Contrôle final : vérification de l'absence de valeurs manquantes dans le jeu de test
# ============================
# Cette vérification confirme que tous les traitements ont été effectués correctement
# et qu'aucune valeur manquante ne subsiste dans le jeu de test nettoyé

na_final_test = test_clean.isna().sum()
na_final_test[na_final_test > 0]
```

```python
# ============================
# Sauvegarde des données nettoyées
# ============================
# Les données nettoyées sont sauvegardées dans le répertoire data/processed
# pour être utilisées dans les étapes suivantes du pipeline (feature engineering, modélisation)

import os

# Création du répertoire de sortie s'il n'existe pas
os.makedirs("data/processed", exist_ok=True)

# Sauvegarde des jeux de données nettoyés au format CSV
# Format CSV choisi pour sa compatibilité avec Kaggle et sa simplicité d'utilisation
train_clean.to_csv("data/processed/train_clean.csv", index=False)
test_clean.to_csv("data/processed/test_clean.csv", index=False)

print(" Fichiers enregistrés :")
print(" - data/processed/train_clean.csv")
print(" - data/processed/test_clean.csv")
```
